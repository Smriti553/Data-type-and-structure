DATA TYPE AND STRUCTURES

1. What are data structures, and why are they important?

->Data structures are ways to organize and store data so it can be used efficiently. They are important because they make operations like searching, sorting, and accessing data faster and easier.

2. Difference between mutable and immutable data types?

->Mutable: Can be changed after creation (e.g., list → [1, 2, 3] → can add/remove elements).
Immutable: Cannot be changed after creation (e.g., string → "hello" cannot be modified directly).

3. Main differences between lists and tuples in Python?

->Lists are mutable, tuples are immutable
Lists use [], tuples use ()
Lists are slower, tuples are faster
Lists are used when data changes, tuples when it should stay fixed

4. How dictionaries store data?

->Dictionaries store data in key-value pairs. Each key is unique and is used to access its corresponding value.

5. Why use a set instead of a list?

->Sets automatically remove duplicates and are faster for checking if an item exists. Useful when you only need unique values.

6. What is a string and how is it different from a list?

->A string is a sequence of characters (e.g., "hello").
Difference:
Strings are immutable, lists are mutable
Strings store characters, lists can store any type of data

7. How tuples ensure data integrity?

->Tuples cannot be changed after creation, so the data remains safe and unchanged throughout the program.

8. What is a hash table and its relation to dictionaries?

->A hash table is a structure that stores data using a hash function. Python dictionaries use hash tables to store and quickly access data using keys.

9. Can lists contain different data types?

->Yes, a list can contain different data types like integers, strings, and even other lists.
Example: [1, "hello", 3.5]

10. Why strings are immutable?

->Strings are immutable to make them faster, more secure, and memory-efficient. It also allows them to be used safely as dictionary keys.

11. Advantages of dictionaries over lists?

->Dictionaries allow fast data retrieval using keys, while lists require searching through elements one by one.

12. How sets handle duplicate values?

->Sets automatically remove duplicates.
Example: {1, 2, 2, 3} becomes {1, 2, 3}

13. When to use a tuple instead of a list?

->Use a tuple when data should not change, like storing coordinates (x, y) or fixed data like dates.

14. How “in” works for lists and dictionaries?

->In a list: checks if a value exists
In a dictionary: checks if a key exists

15. Can you modify elements of a tuple?

->No, because tuples are immutable. Once created, their values cannot be changed.

16. What is a nested dictionary with example?
->A dictionary inside another dictionary.
Example:

student = {
  "name": "Smriti",
  "marks": {"math": 90, "science": 85}
}

17. Time complexity of accessing elements in a dictionary?

->It is usually O(1) (constant time), meaning very fast access using keys.

18. When are lists preferred over dictionaries?

->Lists are preferred when:

Order matters
You need to store simple sequences
You don’t need key-value pairs

19. Why dictionaries are considered unordered?

->Traditionally, dictionaries don’t store data in a fixed order. This means items may not appear in the order they were added (though modern Python preserves insertion order, but conceptually they are unordered).

20. Difference between list and dictionary in data retrieval?

  ->List: Data is accessed using index (position)
  Dictionary: Data is accessed using key (name)
